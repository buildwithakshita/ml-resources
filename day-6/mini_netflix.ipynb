The simplest recommender there is: **K-nearest neighbours**. Every show gets a few scores, which turns it into a dot on a map. To recommend, find the dots closest to a show you loved.

Run each cell with **Shift + Enter**.

```
import pandas as pd
from sklearn.neighbors import NearestNeighbors

# 1. Give every show a score from 0 to 10 for a few things (these are my own guesses!)
shows = pd.DataFrame({
    "show":      ["Friends", "How I Met Your Mother", "New Girl", "The Office", "Brooklyn Nine-Nine", "Seinfeld",
                  "Law & Order", "Breaking Bad", "Succession", "Stranger Things", "The Crown", "Gossip Girl"],
    "funny":     [9, 9, 9, 9, 9, 9, 1, 2, 4, 3, 1, 3],
    "feel_good": [9, 7, 8, 7, 9, 5, 3, 1, 1, 4, 3, 3],
    "romance":   [7, 9, 7, 5, 3, 2, 1, 1, 2, 3, 4, 9],
    "crime":     [0, 0, 0, 0, 6, 0, 10, 10, 3, 3, 0, 1],
    "in_nyc":    [1, 1, 0, 0, 1, 1, 1, 0, 1, 0, 0, 1],
})
shows
```

## The 10 lines that matter

```
features = ["funny", "feel_good", "romance", "crime"]   # the "map"

def recommend(title, cols=features, k=3):
    model = NearestNeighbors(n_neighbors=k + 1).fit(shows[cols])   # learn where every dot is
    i = shows.index[shows["show"] == title][0]                     # find your show
    _, idx = model.kneighbors(shows[cols].iloc[[i]])               # find the closest dots
    return [s for s in shows["show"].iloc[idx[0]] if s != title][:k]

print("Because you watched Friends:", recommend("Friends"))
print("Because you watched Breaking Bad:", recommend("Breaking Bad"))
```

## Add your own show 👀
Change the name and scores, then run it.

```
my_show = {"show": "Ted Lasso", "funny": 8, "feel_good": 10, "romance": 4, "crime": 0, "in_nyc": 0}
shows = pd.concat([shows, pd.DataFrame([my_show])], ignore_index=True)
print("Because you watched", my_show["show"], ":", recommend(my_show["show"]))
```

## The mistake from the reel 😭
The map is only as smart as what you put on it. If the **only** thing we measure is *is it set in New York?*, every NYC show is exactly as close as every other one, so Law & Order counts as a neighbour of Friends.

```
nyc = NearestNeighbors(n_neighbors=len(shows)).fit(shows[["in_nyc"]])
dist, idx = nyc.kneighbors(shows[["in_nyc"]].iloc[[0]])   # row 0 = Friends
same_spot = [s for s, d in zip(shows["show"].iloc[idx[0]], dist[0]) if d == 0 and s != "Friends"]
print("Equally close to Friends when you only track NYC:", same_spot)
```

## What real apps do
Netflix uses much bigger models, but the idea is the same: every show (and every viewer) becomes a long list of numbers that the model **learns** from what millions of people watch, and recommendations come from what's close.
